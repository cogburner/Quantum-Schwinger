```python
from qiskit import QuantumCircuit
import numpy as np
from schwinger_hs_trotter import make_trotter_block_from_y

N, a, g, m, dt = 6, 1.0, 1.0, 0.5, 0.02

# HS sample
rng = np.random.default_rng(7)
y   = rng.standard_normal(N-1)

# One symmetric step (no PauliEvolutionGate)
step = make_trotter_block_from_y(N, y, dt=dt, a=a, g=g, m=m)

# Build total time by composing steps
n_steps = 10
circ = QuantumCircuit(N)
for _ in range(n_steps):
    circ = circ.compose(step, list(range(N)), inplace=False)

print(circ.depth(), "depth; gates:", circ.count_ops())
```
